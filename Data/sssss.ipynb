### Attrition Analysis...

```python
# How much Employee Left the company...

df[df["Attrition"]=="Yes"].shape[0]
```

```python
# Percentage of Attrition...

employees_left = df[df["Attrition"] == "Yes"].shape[0]
total_employees = len(df)
attrition_rate = (
    employees_left / total_employees
) * 100

print("Attrition Rate:", round(attrition_rate,2), "%")
```

```python
df["Attrition"].value_counts().plot(
    kind="bar"
)

plt.title("Attrition Count")
plt.xlabel("Attrition")
plt.ylabel("Employees")

plt.show()
```

### Departmental Analysis...

```python
df.groupby("Department")["Attrition"].value_counts()
```

```python
df["Department"].value_counts().plot(
    kind="bar"
)

plt.title("Employees by Department")

plt.show()
```

```python
# Department wise Attrition...

department_attrition = pd.crosstab(
    df["Department"],
    df["Attrition"]
)

department_attrition
```

### Gender Analysis...

```python
df["Gender"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")

plt.title("Gender Distribution")

plt.show()
```

```python
# Gender wise attrition...

pd.crosstab(
    df["Gender"],
    df["Attrition"]
)
```

### Salary Analysis...
